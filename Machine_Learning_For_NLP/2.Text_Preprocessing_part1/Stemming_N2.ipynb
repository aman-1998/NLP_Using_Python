#### Stemming:
Stemming is a text preprocessing technique in Natural Language Processing (NLP) that reduces words to their root-like form by removing prefixes or suffixes.

<u>Example of Stemming:</u>
<br>playing => play
<br>played => play
<br>plays => play
<br>studies	=> studi
<br>studying => studi
<br>connection => connect
<br>history => histori

Notice that studies becomes studi, which is not a proper English word. Because stemming generally applies rules to remove word endings rather than understanding the meaning of each word.

```python
from nltk.stem import PorterStemmer

words = ["programs", "running", "jumps", "connection", "history", "studying", "study", "congratulations", "generously"]

ps = PorterStemmer()

for word in words:
    print(ps.stem(word))
```

```python
from nltk.stem import RegexpStemmer

words = ["programs", "running", "jumps", "connection", "history", "studying", "study", "congratulations", "generously"]

rs = RegexpStemmer('ing$|s$|ed$|ion$|ations$')

for word in words:
    print(rs.stem(word))
```

```python
from nltk.stem import SnowballStemmer

words = ["programs", "running", "jumps", "connection", "history", "studying", "study", "congratulations", "generously"]

ss = SnowballStemmer("english")

for word in words:
    print(ss.stem(word))
    
# SnowballStemmer is more aggressive than PorterStemmer and handles a wider range of word forms.
# Example: PorterStemmer: generously -> gener
#          SnowballStemmer: generously -> generous
```
