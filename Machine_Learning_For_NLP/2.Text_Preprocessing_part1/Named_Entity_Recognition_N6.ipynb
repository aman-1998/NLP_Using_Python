```python
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
from nltk import pos_tag, ne_chunk

sentence = "Shubman Gill (born September 8, 1999) is an Indian international cricketer and the captain of the Indian Test and ODI cricket teams."

tokens = word_tokenize(sentence)

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
    
filtered_tokens=[]
for token in tokens:
    if token.isalnum() and token.lower() not in stop_words:
        filtered_tokens.append(token)

# POS tagging
tagged_tokens = pos_tag(filtered_tokens)

print(tagged_tokens)

named_entity = ne_chunk(tagged_tokens)

print(named_entity)

named_entity.draw()
```

In the above example, preprocessing removes punctuation and stop words before NER (Named Entity Recognition). Instead, perform NER on the original tokens .i.e., don't remove stop words and lower the case.

For better results in case of NER, we should give as much information as possible (don't remove stopwords, etc)

```python
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
from nltk import pos_tag, ne_chunk

sentence = "Shubman Gill (born September 8, 1999) is an Indian international cricketer and the captain of the Indian Test and ODI cricket teams."

tokens = word_tokenize(sentence)

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
    
tokens = word_tokenize(sentence)

tagged_tokens = pos_tag(tokens)

print(tagged_tokens)

named_entity = ne_chunk(tagged_tokens)

print(named_entity)
named_entity.draw()
```
