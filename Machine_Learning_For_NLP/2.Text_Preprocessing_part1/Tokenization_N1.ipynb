### <u>Important terms used in NLP:</u>
#### Corpus => Paragraph
#### Document => Sentence
#### Vocabulary => Unique words
#### Words => words

#### Tokenization means breaking a corpus into sentences or breaking a sentence into words.

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize

text1 = "Hello world!! This is a test sentence. Tokenization is fun!"
listOfDocuments1 = sent_tokenize(text1)
print(listOfDocuments1)

print("=================================================================================")

text2 = '''Virat Kohli is an Indian international cricketer and former captain of the 
Indian national cricket team, widely regarded as one of the greatest batters 
in cricket history. Known as "King Kohli" he is a dominant right-handed 
batsman who completely revolutionized the modern era of the sport through his 
unmatched standard of batting technique, chasing ability, and extreme fitness.
'''
listOfDocuments2 = sent_tokenize(text2)
for i,sentence in enumerate(listOfDocuments2):
    print(f"Sentence {i+1}: {sentence}")
```

```python
from nltk.tokenize import word_tokenize

text = "Hello world!! This is a test sentence. Tokenization is fun! I am watching Krish Naik's NLP Tutorial."
words = word_tokenize(text)
print(words)

listOfDocuments = sent_tokenize(text)
listOfWords = []
for sentence in listOfDocuments:
    words_in_sentence = word_tokenize(sentence)
    listOfWords.extend(words_in_sentence)
    
print(listOfWords)
```

```python
from nltk.tokenize import wordpunct_tokenize

text = "Hello world!! This is a test sentence. Tokenization is fun! I am watching Krish Naik's NLP Tutorial."
words = wordpunct_tokenize(text)
print(words)

# Notice how wordpunct_tokenize splits punctuation into separate tokens

# For example, word_tokenize would keep "Hello world!!" as ["Hello", "world", "!", "!"]
# wordpunct_tokenize would keep "Hello world!!" as ["Hello", "world", "!!"]

# For word_tokenize, Naik's would be split into ["Naik", "'s"]
# For wordpunct_tokenize, Naik's would be split into ["Naik", "'", "s"]
```

```python
from nltk.tokenize import TreebankWordTokenizer

text = "Hello world!! This is a test sentence. Tokenization is fun! I am watching Krish Naik's NLP Tutorial."
tokenizer = TreebankWordTokenizer()
words = tokenizer.tokenize(text)
print(words)

# TreebankWordTokenizer appends full-stops with the word itself rather than treating them as separate tokens.
# But last full-stop is treated as a separate token.
# For instance, in the sentence above, the final full-stop after "Tutorial" is treated as a separate token.
```
