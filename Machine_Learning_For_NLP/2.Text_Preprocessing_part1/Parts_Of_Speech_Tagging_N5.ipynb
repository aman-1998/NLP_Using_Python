````python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
from nltk import pos_tag

paragraph = '''
Virat Kohli (born 5 November 1988) is an Indian international cricketer and former all-format captain of the Indian national cricket team. He is a right-handed batter and an occasional right-arm medium-pace bowler. Considered one of the greatest batters in cricket, he has been acclaimed for his batting skills and records. Kohli has the most centuries in ODIs and the second-most centuries in international cricket with 86 tons across all formats. He is also the leading run-scorer in the Indian Premier League.

Kohli was the captain of the 2008 U19 World Cup winning team and was a crucial member of the Indian teams that won 2011 Cricket World Cup, 2013 Champions Trophy, 2024 T20 World Cup, and 2025 Champions Trophy. He plays for Royal Challengers Bengaluru in the Indian Premier League and for Delhi in domestic cricket. In 2013, Kohli was ranked number one in the ODI batting rankings. In 2015, he achieved the same in T20I cricket. In 2018, he was ranked number one in Test cricket, making him the only Indian to hold the top spot in all three formats. He is the first player to score 20,000 international runs in a decade. He was the Cricketer of the Decade for 2011 to 2020.
'''

documents = sent_tokenize(paragraph)

lemmatizer = WordNetLemmatizer()

pos_for_each_words = []
for doc in documents:
    tokens = word_tokenize(doc)
    filtered_tokens = []
    for token in tokens:
        if token.isalpha() and token.lower() not in stopwords.words('english'):
            lemmatized_token = lemmatizer.lemmatize(token, 'v')
            filtered_tokens.append(lemmatized_token)
    pos_for_each_words.append(pos_tag(filtered_tokens))

# processed documents after stopword removal and lemmatization
for i,doc in enumerate(pos_for_each_words):
    print(f"sentence {i}: {doc}")
````

The error occurs because NLTK's pos_tag() and WordNetLemmatizer() use different POS tag formats.

Your code passes a Penn Treebank POS tag such as 'NN' directly to the lemmatizer:

```
pos = pos_tag([token.lower()])[0][1]
lemmatized_token = lemmatizer.lemmatize(token.lower(), pos)
```

However, WordNetLemmatizer expects WordNet POS tags, not Penn Treebank tags.

````python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords, wordnet
from nltk import pos_tag

paragraph = '''
Virat Kohli (born 5 November 1988) is an Indian international cricketer and former all-format captain of the Indian national cricket team. He is a right-handed batter and an occasional right-arm medium-pace bowler. Considered one of the greatest batters in cricket, he has been acclaimed for his batting skills and records. Kohli has the most centuries in ODIs and the second-most centuries in international cricket with 86 tons across all formats. He is also the leading run-scorer in the Indian Premier League.

Kohli was the captain of the 2008 U19 World Cup winning team and was a crucial member of the Indian teams that won 2011 Cricket World Cup, 2013 Champions Trophy, 2024 T20 World Cup, and 2025 Champions Trophy. He plays for Royal Challengers Bengaluru in the Indian Premier League and for Delhi in domestic cricket. In 2013, Kohli was ranked number one in the ODI batting rankings. In 2015, he achieved the same in T20I cricket. In 2018, he was ranked number one in Test cricket, making him the only Indian to hold the top spot in all three formats. He is the first player to score 20,000 international runs in a decade. He was the Cricketer of the Decade for 2011 to 2020.
'''

documents = sent_tokenize(paragraph)
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

# Convert Penn Treebank POS tags to WordNet POS tags
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

filtered_documents = []
for doc in documents:
    tokens = word_tokenize(doc)
    filtered_tokens = []
    for token in tokens:
        if token.isalnum() and token.lower() not in stop_words:
            treebank_pos = pos_tag([token.lower()])[0][1]
            lemmatized_token = lemmatizer.lemmatize(token, get_wordnet_pos(treebank_pos))
            filtered_tokens.append(lemmatized_token)
    filtered_documents.append(" ".join(filtered_tokens))

# processed documents after stopword removal and lemmatization
for i,doc in enumerate(filtered_documents):
    print(f"sentence {i}: {doc}")
````

````python
# A better approach than above

from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords, wordnet
from nltk import pos_tag

paragraph = '''
Virat Kohli (born 5 November 1988) is an Indian international cricketer and former all-format captain of the Indian national cricket team. He is a right-handed batter and an occasional right-arm medium-pace bowler. Considered one of the greatest batters in cricket, he has been acclaimed for his batting skills and records. Kohli has the most centuries in ODIs and the second-most centuries in international cricket with 86 tons across all formats. He is also the leading run-scorer in the Indian Premier League.

Kohli was the captain of the 2008 U19 World Cup winning team and was a crucial member of the Indian teams that won 2011 Cricket World Cup, 2013 Champions Trophy, 2024 T20 World Cup, and 2025 Champions Trophy. He plays for Royal Challengers Bengaluru in the Indian Premier League and for Delhi in domestic cricket. In 2013, Kohli was ranked number one in the ODI batting rankings. In 2015, he achieved the same in T20I cricket. In 2018, he was ranked number one in Test cricket, making him the only Indian to hold the top spot in all three formats. He is the first player to score 20,000 international runs in a decade. He was the Cricketer of the Decade for 2011 to 2020.
'''

documents = sent_tokenize(paragraph)

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

# Convert Penn Treebank POS tags to WordNet POS tags
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

filtered_documents = []

for doc in documents:
    tokens = word_tokenize(doc)

    # Remove stop words before POS tagging
    filtered_tokens=[]
    for token in tokens:
        if token.isalnum() and token.lower() not in stop_words:
            filtered_tokens.append(token)

    # POS tagging
    tagged_tokens = pos_tag(filtered_tokens)

    # Lemmatization using converted POS tags
    lemmatized_tokens = []

    for token, tag in tagged_tokens:
        wordnet_pos = get_wordnet_pos(tag)
        lemmatized_token = lemmatizer.lemmatize(token, wordnet_pos)
        lemmatized_tokens.append(lemmatized_token)

    # Store POS tags for lemmatized words
    filtered_documents.append(" ".join(lemmatized_tokens))

# Print processed documents
for i, doc in enumerate(filtered_documents):
    print(f"Sentence {i}: {doc}")
````
