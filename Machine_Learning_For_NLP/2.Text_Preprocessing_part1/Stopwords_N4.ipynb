#### Stopwords

Stop words are common words that are often removed from text during preprocessing because they may contribute little to a particular NLP task.

Examples include is, am, are, the, a, an, in, on, of, and, and the.

However, stop words are not always meaningless. Whether to remove them depends on the task.

Example:

<span style="color: red;">Original sentence:</span> <b>The boy is playing cricket in the park.</b>

Suppose we remove common stop words such as the, is, and in.

<span style="color: red;">After stop-word removal:</span> <b>boy playing cricket park</b>

The sentence becomes shorter, retaining many of its main content words.

```python
from nltk.tokenize import word_tokenize
from nltk.stem import SnowballStemmer
from nltk.corpus import stopwords

paragraph = '''
Virat Kohli (born 5 November 1988) is an Indian international cricketer and former all-format captain of the Indian national cricket team. He is a right-handed batter and an occasional right-arm medium-pace bowler. Considered one of the greatest batters in cricket, he has been acclaimed for his batting skills and records. Kohli has the most centuries in ODIs and the second-most centuries in international cricket with 86 tons across all formats. He is also the leading run-scorer in the Indian Premier League.

Kohli was the captain of the 2008 U19 World Cup winning team and was a crucial member of the Indian teams that won 2011 Cricket World Cup, 2013 Champions Trophy, 2024 T20 World Cup, and 2025 Champions Trophy. He plays for Royal Challengers Bengaluru in the Indian Premier League and for Delhi in domestic cricket. In 2013, Kohli was ranked number one in the ODI batting rankings. In 2015, he achieved the same in T20I cricket. In 2018, he was ranked number one in Test cricket, making him the only Indian to hold the top spot in all three formats. He is the first player to score 20,000 international runs in a decade. He was the Cricketer of the Decade for 2011 to 2020.
'''

tokens = word_tokenize(paragraph)
ss = SnowballStemmer('english')

# Remove stopwords
filtered_tokens = []
for token in tokens:
    if token.lower() not in stopwords.words('english'):
        stemmed_token = ss.stem(token.lower())
        filtered_tokens.append(stemmed_token)

print(filtered_tokens)
```

```python
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords

paragraph = '''
Virat Kohli (born 5 November 1988) is an Indian international cricketer and former all-format captain of the Indian national cricket team. He is a right-handed batter and an occasional right-arm medium-pace bowler. Considered one of the greatest batters in cricket, he has been acclaimed for his batting skills and records. Kohli has the most centuries in ODIs and the second-most centuries in international cricket with 86 tons across all formats. He is also the leading run-scorer in the Indian Premier League.

Kohli was the captain of the 2008 U19 World Cup winning team and was a crucial member of the Indian teams that won 2011 Cricket World Cup, 2013 Champions Trophy, 2024 T20 World Cup, and 2025 Champions Trophy. He plays for Royal Challengers Bengaluru in the Indian Premier League and for Delhi in domestic cricket. In 2013, Kohli was ranked number one in the ODI batting rankings. In 2015, he achieved the same in T20I cricket. In 2018, he was ranked number one in Test cricket, making him the only Indian to hold the top spot in all three formats. He is the first player to score 20,000 international runs in a decade. He was the Cricketer of the Decade for 2011 to 2020.
'''

tokens = word_tokenize(paragraph)

lemmatizer = WordNetLemmatizer()

filtered_tokens = []

# Lemmatization performs better than snowball stemmer for verbs but takes more computational time.
for token in tokens:
    if token.lower() not in stopwords.words('english'):
        lemmatized_token = lemmatizer.lemmatize(token.lower(), 'v')
        filtered_tokens.append(lemmatized_token)

print(filtered_tokens)
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords

paragraph = '''
Virat Kohli (born 5 November 1988) is an Indian international cricketer and former all-format captain of the Indian national cricket team. He is a right-handed batter and an occasional right-arm medium-pace bowler. Considered one of the greatest batters in cricket, he has been acclaimed for his batting skills and records. Kohli has the most centuries in ODIs and the second-most centuries in international cricket with 86 tons across all formats. He is also the leading run-scorer in the Indian Premier League.

Kohli was the captain of the 2008 U19 World Cup winning team and was a crucial member of the Indian teams that won 2011 Cricket World Cup, 2013 Champions Trophy, 2024 T20 World Cup, and 2025 Champions Trophy. He plays for Royal Challengers Bengaluru in the Indian Premier League and for Delhi in domestic cricket. In 2013, Kohli was ranked number one in the ODI batting rankings. In 2015, he achieved the same in T20I cricket. In 2018, he was ranked number one in Test cricket, making him the only Indian to hold the top spot in all three formats. He is the first player to score 20,000 international runs in a decade. He was the Cricketer of the Decade for 2011 to 2020.
'''

documents = sent_tokenize(paragraph)

lemmatizer = WordNetLemmatizer()

filtered_documents = []
for doc in documents:
    tokens = word_tokenize(doc)
    filtered_tokens = []
    for token in tokens:
        if token.lower() not in stopwords.words('english'):
            lemmatized_token = lemmatizer.lemmatize(token.lower(), 'v')
            filtered_tokens.append(lemmatized_token)
    filtered_documents.append(filtered_tokens)

# processed documents after stopword removal and lemmatization
for doc in filtered_documents:
    print(doc)
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords

paragraph = '''
Virat Kohli (born 5 November 1988) is an Indian international cricketer and former all-format captain of the Indian national cricket team. He is a right-handed batter and an occasional right-arm medium-pace bowler. Considered one of the greatest batters in cricket, he has been acclaimed for his batting skills and records. Kohli has the most centuries in ODIs and the second-most centuries in international cricket with 86 tons across all formats. He is also the leading run-scorer in the Indian Premier League.

Kohli was the captain of the 2008 U19 World Cup winning team and was a crucial member of the Indian teams that won 2011 Cricket World Cup, 2013 Champions Trophy, 2024 T20 World Cup, and 2025 Champions Trophy. He plays for Royal Challengers Bengaluru in the Indian Premier League and for Delhi in domestic cricket. In 2013, Kohli was ranked number one in the ODI batting rankings. In 2015, he achieved the same in T20I cricket. In 2018, he was ranked number one in Test cricket, making him the only Indian to hold the top spot in all three formats. He is the first player to score 20,000 international runs in a decade. He was the Cricketer of the Decade for 2011 to 2020.
'''

documents = sent_tokenize(paragraph)

lemmatizer = WordNetLemmatizer()

filtered_documents = []
for doc in documents:
    tokens = word_tokenize(doc)
    filtered_tokens = []
    for token in tokens:
        if token.lower() not in stopwords.words('english'):
            lemmatized_token = lemmatizer.lemmatize(token.lower(), 'v')
            filtered_tokens.append(lemmatized_token)
    filtered_documents.append(" ".join(filtered_tokens))

# processed documents after stopword removal and lemmatization
for i,doc in enumerate(filtered_documents):
    print(f"sentence {i}: {doc}")
```
