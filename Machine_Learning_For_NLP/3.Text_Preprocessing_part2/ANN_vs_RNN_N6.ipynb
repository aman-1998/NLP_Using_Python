#### ANN vs RNN

ANN = Artificial Neural Network
<br>RNN = Recurant Neural Network

ANN is used when we don't need memory to remember previous inputs i.e., result doesn't depends on continuous flow of data (or previous data)

RNN is used when there is a continuous flow of data and we need to memory to store previous data to generate future output.

<u><b>Activation function:</b></u>
An activation function is a mathematical function used in a neural network to determine the output of a neuron.

In simple terms, it decides how strongly a neuron should activate based on its calculated input.

The output that we get in the hidden layer or output layer are passed through the activation function to get the final output.

<u>Common activation functions:</u>

A. <u>Sigmoid Function:</u> 

f(x) = 1 / 1+e<sup>-x</sup>

=> f'(x) = f(x).(1 - f(x))

- Output range: 0 to 1.
- Commonly used in binary classification output layers.

B. <u>Tanh:</u>

f(x) = tanh x = (e<sup>x</sup>-e<sup>-x</sup>) / (e<sup>x</sup>+e<sup>-x</sup>)

=> f'(x) = 1 - (f(x))<sup>2</sup>

- Output range: -1 to +1.
- Output is centred around zero.

C. <u>ReLU (Rectified Linear Unit):</u>

```
f(x) = max(0,x)
```

```
f(x) = x when x > 0
       0 when x <= 0 

=> f'(x) = 1 when x > 0
           0 when x <= 0
```
       
- If input is negative, output is 0.
- If input is positive, output remains unchanged.
- Commonly used in hidden layers.

D. <u>Leaky ReLU (Leaky Rectified Linear Unit):</u>

```
f(x) = x when x > 0
       α.x when x <= 0, where α is a small constant

=> f'(x) = 1 when x > 0
           α when x <= 0
```

- If input is negative, output is a small negative value.
- If input is positive, output remains unchanged.
- Commonly used in hidden layers.

Remember:
- ReLU: Negative input → 0.
- Leaky ReLU: Negative input → Small negative value.

The main purpose of Leaky ReLU is to reduce the Dying ReLU problem by allowing a small gradient for negative inputs.

E. <u>Softmax:</u>                 

f(x<sub>i</sub>) = e<sup>x<sub>i</sub></sup> /  ∑<sup>n</sup><sub>j=1</sub> e<sup>x<sub>j</sub></sup>

x<sub>i</sub>: Individual input score (logit).
<br>e: Euler's number (approximately 2.71828).
<br>n: Total number of classes.
<br>Denominator: Sum of the exponential values of all logits.

Example:
Suppose a neural network identifies three fruits:

| Fruit  | Logit |
|--------|-------|
| Apple  |   2   |
| Banana |   1   |
| Orange |   0.5 |

<b>Step 1:</b> Apply exponential (e) to each logit.

e<sup>2</sup>=7.389

e<sup>1</sup>=2.718

e<sup>0.5</sup>=1.649

<b>Step 2:</b> Calculate the sum.

7.389+2.718+1.649=11.756

<b>Step 3:</b> Apply the Softmax formula.

P(Apple) = 7.389 / 11.756 = 0.6285

P(Banana)= 2.718 / 11.756 = 0.2312

P(Orange)=1.649 / 11.756 = 0.1403

Final result:

| Fruit     | Probability |
|-----------|-------------|
| Apple     |   62.85%    |
| Banana    |   23.12%    |
| Orange    |   14.03%    |
| **Total** | **100%**    |

The model predicts Apple because it has the highest Softmax output.

<b>Loss Function:</b>

A Loss Function is a mathematical function that measures how far a neural network's prediction is from the actual (correct) answer.

In simple terms, a loss function tells the neural network how wrong its prediction is. The network uses this information to improve its predictions during training.

Different machine learning problems require different loss functions.

| Loss function                     | Common application                            |
|-----------------------------------|-----------------------------------------------|
| Mean Squared Error (MSE)          | Regression                                    |
| Mean Absolute Error (MAE)         | Regression                                    |
| Binary Cross-Entropy              | Binary classification                         |
| Categorical Cross-Entropy         | Multiclass classification                     |
| Sparse Categorical Cross-Entropy  | Multiclass classification with integer labels |
| Hinge Loss                        | Support Vector Machines                       |

<u><b>Different types of Loss Function:</b></u>

<u><b>1. Regression Loss Functions</b></u>

Regression means predicting a continuous numerical value.

Examples:
- Predicting house prices.
- Predicting someone's salary.
- Predicting temperature.
- Predicting stock prices.

<u><b>A. Mean Squared Error (MSE)</b></u>

MSE calculates the average of the squared differences between actual and predicted values.

MSE = (1/N) . ∑<sup>n</sup><sub>i=1</sub> (y<sub>i</sub>-ŷ<sub>i</sub>)<sup>2</sup>

y<sub>i</sub> = Actual value
<br>ŷ<sub>i</sub> = Predicted value
<br>N = Number of observations

<u>Limitation:</u> MSE is sensitive to outliers because large errors get squared.


````python
from sklearn.metrics import mean_squared_error

actual = [50, 60, 70]
predicted = [55, 58, 80]

print(mean_squared_error(actual, predicted))
````

<u><b>B. Mean Absolute Error (MAE)</b></u>

MAE calculates the average absolute difference between actual and predicted values.

MAE = (1/N) . ∑<sup>n</sup><sub>i=1</sub> |y<sub>i</sub>-ŷ<sub>i</sub>|

y<sub>i</sub> = Actual value
<br>ŷ<sub>i</sub> = Predicted value
<br>N = Number of observations

Why use MAE?
- It is less sensitive to outliers than MSE.
- Its output is in the same unit as the target variable.
- It gives a straightforward interpretation of average absolute error.

````python
from sklearn.metrics import mean_absolute_error

actual = [50, 60, 70]
predicted = [55, 58, 80]

print(mean_absolute_error(actual, predicted))
````

<u><b>C. Root Mean Squared Error (RMSE)</b></u>

RMSE is simply the square root of MSE.

RMSE = sqrt( (1/N) . ∑<sup>n</sup><sub>i=1</sub> (y<sub>i</sub>-ŷ<sub>i</sub>)<sup>2</sup> )

Why use RMSE?

MSE produces squared units, whereas RMSE brings the result back to the original unit.

For example, if we predict house prices in lakhs, RMSE is also measured in lakhs.

````python
import numpy as np

actual = [50, 60, 70]
predicted = [55, 58, 80]

rmse = np.sqrt(mean_squared_error(actual, predicted))

print(rmse)
````

<u><b>D. Huber Loss</b></u>

Huber Loss combines the characteristics of MSE and MAE.
- For smaller errors, it behaves like MSE.
- For larger errors, it behaves like MAE.


L<sub>δ</sub>(x) = (y<sub>i</sub>-ŷ<sub>i</sub>)<sup>2</sup> / 2
<br>&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;δ(|y<sub>i</sub>-ŷ<sub>i</sub>| - δ/2)

δ is a threshold controlling when the loss changes from quadratic to linear.

Why use it?

It is useful when your dataset contains outliers but you still want the smooth behaviour of MSE around small errors.


<u><b>2. Classification Loss Functions</b></u>

Classification means predicting a category or class.

Examples:
- Spam or Ham.
- Cat, Dog or Horse.
- Apple, Banana or Orange.
- Fraudulent or legitimate transaction.

<u><b>A. Binary Cross-Entropy (BCE) or Log Loss</b></u>

Used when there are two mutually exclusive classes.

Example:
- Spam = 1
- Ham = 0

L=-[y.log(p) + (1-y).log(1-p)]

Where:
- y = Actual label (0 or 1).
- p = Predicted probability of class 1.

Example:

Suppose an email is actually Spam.

y=1

The model predicts:

p=0.8

Therefore:

L=-[1.log(0.8) + 0.log(0.2)]

L=0.2231

Now suppose the model predicts only 20% probability of Spam:

L=-log(0.2)

L=1.6094

Notice that the loss increases when the model assigns a low probability to the correct class.


````python
from sklearn.metrics import log_loss

actual = [1, 0, 1]
predicted = [0.8, 0.3, 0.6]

print(log_loss(actual, predicted))
````

<u><b>B. Categorical Cross-Entropy (CCE)</b></u>

Used for multiclass classification when actual labels are one-hot encoded.

Let's use our previous fruit example.

| Fruit   | Actual label | Predicted probability |
|---------|--------------|-----------------------|
| Apple   | 1            | 0.60                  |
| Banana  | 0            | 0.30                  |
| Orange  | 0            | 0.10                  |

L = - ∑<sup>n</sup><sub>i=1</sub> y<sub>i</sub>.log(p<sub>i</sub>)

````python
from tensorflow.keras.losses import CategoricalCrossentropy

loss_fn = CategoricalCrossentropy()

actual = [[1, 0, 0]]
predicted = [[0.60, 0.30, 0.10]]

print(loss_fn(actual, predicted).numpy())
````

<u><b>C. Sparse Categorical Cross-Entropy</b></u>

It is mathematically the same as Categorical Cross-Entropy.

The difference is how actual labels are represented.

Let's use our previous fruit example.

Suppose we have three possible fruits:
- Apple → Class 0
- Banana → Class 1
- Orange → Class 2


| Fruit   | Class Index   | Actual label | Predicted probability |
|---------|---------------|--------------|-----------------------|
| Apple   |     0         | 1            | 0.60                  |
| Banana  |     1         | 0            | 0.30                  |
| Orange  |     2         | 0            | 0.10                  |

L = - ∑<sup>n</sup><sub>i=1</sub> y<sub>i</sub>.log(p<sub>i</sub>)

Instead of creating a vector, we simply provide the class index of the correct answer.

<b>Categorical Cross-Entropy:</b> actual = [[1, 0, 0]]

<b>Sparse Categorical Cross-Entropy:</b> actual = [0]

````python
import tensorflow as tf
from tensorflow.keras.losses import SparseCategoricalCrossentropy

actual = tf.constant([0], dtype=tf.int32)

predicted = tf.constant([[0.60, 0.30, 0.10]], dtype=tf.float32)

loss_fn = SparseCategoricalCrossentropy()

loss = loss_fn(actual, predicted)

print(loss.numpy())
````

<u><b>D. Hinge Loss</b></u>

Hinge Loss is a loss function primarily used in Support Vector Machines (SVMs) for classification.

Hinge Loss is commonly associated with Support Vector Machines (SVMs).

Unlike Cross-Entropy, which evaluates predicted probabilities, Hinge Loss focuses on:
1. Whether the prediction is correct.
2. How confidently the model separates the classes.
3. Whether the prediction has a sufficient safety margin.
Let's understand it step by step.

L = max(0, 1-y.f(x))

Where:
- y = Actual class label (+1 or -1).
- f(x) = Model's prediction score.

Let's understand it step by step.

1. First, understand binary classification

Suppose we are building an email spam classifier.

We have two possible classes:
Email	Class label
Spam	+1
Not Spam (Ham)	-1


Notice something important: Hinge Loss uses labels -1 and +1, rather than 0 and 1.

Our model produces a score called the decision score:
- Positive score → Spam.
- Negative score → Ham.
- Score close to zero → Model is uncertain about the classification.

| Email | Actual label | Model score | Prediction |
|---|---:|---:|---|
| A | +1 | +3.5 | Spam |
| B | +1 | +0.5 | Spam |
| C | -1 | -2.5 | Ham |
| D | -1 | +1.5 | Spam |

Notice:
- Email A is correctly classified with a strong positive score.
- Email B is correctly classified but with a relatively weak score.
- Email C is correctly classified with a strong negative score.
- Email D is incorrectly classified.

**Case 1: Correct prediction with sufficient margin**

Suppose:
- Actual label = +1 (Spam)
- Model score = +3.5

Applying the formula:

L=max(0,1-(+1 x 3.5))

L=max(0,1-3.5)

L=max(0,-2.5)

L=0

Why is the loss zero?

Because the model correctly classified the email and its decision score exceeded the required margin of 1.

The model has satisfied the margin requirement.

**Case 2: Correct prediction but insufficient margin**

Suppose:
- Actual label = +1 (Spam)
- Model score = +0.5

Applying the formula:

L=\max(0,1-(+1 x 0.5))

L=max(0,0.5)

L=0.5

The prediction is correct because the score is positive.

However, the score is only 0.5, whereas the required margin is 1.

Therefore, the model receives a loss of 0.5.

Important: Hinge Loss penalizes even correctly classified samples when their margin is insufficient.

**Case 3: Incorrect prediction**

Suppose:
- Actual label = -1 (Ham)
- Model score = +1.5

Applying the formula:

L=max(0,1-(-1 x 1.5))

L=max(0,1+1.5)

L=2.5

The actual email is Ham, but the model predicts Spam.

Therefore, the model receives a loss of 2.5.

The incorrect prediction violates the margin requirement.

**What happens when the actual label is -1?**

Let's understand why we multiply the actual label by the model score.

The formula is:

L=max(0,1-y.f(x))

When the actual label is +1:

L=max(0,1-f(x))

The model needs a sufficiently positive score.

When the actual label is -1:

L=max(0,1+f(x))

The model needs a sufficiently negative score.

This is how the same formula handles both classes.

**Complete comparison:**

| Actual label (y) | Model score f(x) | Prediction | Hinge Loss |
|---:|---:|---|---:|
| +1 | +3.5 | Correct | 0 |
| +1 | +0.5 | Correct, insufficient margin | 0.5 |
| +1 | -2.0 | Incorrect | 3 |
| -1 | -3.0 | Correct | 0 |
| -1 | -0.5 | Correct, insufficient margin | 0.5 |
| -1 | +1.5 | Incorrect | 2.5 |


Notice the pattern:
- Correct prediction with sufficient margin → Zero loss.
- Correct prediction without sufficient margin → Positive loss.
- Incorrect prediction → Positive loss.

Example:

| Actual | Predicted score | Hinge Loss |
|---:|---:|---:|
| +1 | +3.5 | 0 |
| +1 | +0.5 | 0.5 |
| -1 | -2.5 | 0 |
| -1 | +1.5 | 2.5 |

Average Loss = (0+0.5+0+2.5) /4 = 0.75

````python
import numpy as np

def hinge_loss(actual, predicted):
    loss = np.maximum(0, 1 - actual * predicted)
    return np.mean(loss)

actual = np.array([1, 1, -1, -1])

predicted = np.array([3.5, 0.5, -2.5, 1.5])

loss = hinge_loss(actual, predicted)

print(loss)
````
