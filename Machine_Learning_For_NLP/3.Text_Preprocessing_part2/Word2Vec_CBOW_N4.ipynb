In natural language processing, a **word embedding** is a representation of a word. The embedding is used in text analysis. Typically, the representation is a real-valued vector that encodes the meaning of the word in such a way that the words that are closer in the vector space are expected to be similar in meaning.

Word Embeddings is of two types.
1. Based on count and frequency of words
2. Deep Learning Trained Models

**One Hot Encoding**, **Bag Of Words**, **TFIDF**, etc are example of frequency based word embeddings.

**Word2Vec** comes under Deep Learning based word embeddings.

Word2Vec is of two types:
1. Continuous Bag Of Words (CBOW)
2. SkipGram

In a 2D-vector space, if we plot Banana and Apple coordinates then they will be very close to each other.

```python
from IPython.display import SVG, display

svg = """
<svg width="500" height="450" viewBox="0 0 500 450">

    <!-- X-axis -->
    <line x1="40" y1="300" x2="460" y2="300"
          stroke="orange" stroke-width="4"/>
    <polygon points="460,300 445,291 445,309" fill="orange"/>

    <!-- Y-axis -->
    <line x1="40" y1="300" x2="40" y2="20"
          stroke="orange" stroke-width="4"/>
    <polygon points="40,20 31,35 49,35" fill="orange"/>

    <!-- Apple vector -->
    <line x1="40" y1="300" x2="160" y2="150"
          stroke="lime" stroke-width="3"/>
    <circle cx="160" cy="150" r="5" fill="red"/>

    <!-- Banana vector -->
    <line x1="40" y1="300" x2="190" y2="175"
          stroke="lime" stroke-width="3"/>
    <circle cx="190" cy="175" r="5" fill="red"/>

    <!-- Labels -->
    <text x="145" y="130" fill="blue" font-size="25">Apple</text>
    <text x="210" y="200" fill="blue" font-size="25">Banana</text>

    <!-- Angle arc: between the two vectors -->
    <path d="M 75.5 255.6 A 50 50 0 0 1 80 266.7"
          fill="none" stroke="orange" stroke-width="2"/>

    <!-- Theta -->
    <text x="100" y="250" fill="orange" font-size="20">θ</text>

    <!-- Distance formula -->
    <text x="50" y="380" fill="blue" font-size="25">
        Distance = 1 - cosine_similarity
    </text>
    <text x="162" y="415" fill="blue" font-size="25">
        = 1 - cosθ
    </text>

</svg>
"""

display(SVG(svg))
```

If θ = 90 then cos(90) = 0 
<br>=> Distance = 1 - 0 = 1

If θ = 45 then cos(45) = 0.7071 
<br>=> Distance = 1 - 0.7071 = 0.3

If θ = 0 then cos(0) = 1 
<br>=> Distance = 1 - 1 = 0

If θ = 180 then cos(180) = -1 
<br>=> Distance = 1 - (-1) = 2
